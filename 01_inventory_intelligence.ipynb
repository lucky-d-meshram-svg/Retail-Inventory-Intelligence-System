##### 1.Revenue Analysis :
 
Revenue measures the total income generated from product sales.
 
###### Formula: Revenue = Sales Quantity × Price

```python
demand["revenue"] = (
    demand["Sales_Quantity"] * demand["Price_"]
    )
```

```python
demand[["Sales_Quantity", "Price_", "revenue"]].head()
```

```python
total_revenue = demand["revenue"].sum()
print("Total Revenue",total_revenue)
```

##### 2.Inventory Risk Analysis :
This KPI identifies products that may run out of stock.
 
###### Rule: Stock Level < Reorder Point = High Risk

```python
inventory["Inventory_Risk"] = inventory.apply(
    lambda x: "High Risk"
    if x["Stock_Levels"] < x["Reorder_Point"]
    else "Safe",
    axis=1
)

inventory[["Product_ID","Stock_Levels","Reorder_Point","Inventory_Risk"]].head()
```

##### 3. Price Difference Analysis :
 
This KPI compares product prices with competitor prices.
 
###### Formula: Price Difference = Product Price − Competitor Price

```python
pricing["Price_Difference"] = (
    pricing["Price_"] -
    pricing["Competitor_Prices"]
)

pricing[
    ["Product_ID", "Price_", "Competitor_Prices", "Price_Difference"]
].head()
```

##### 4. Inventory Coverage Analysis :
 
This KPI measures how much inventory is available compared to the reorder point.
 
###### Formula: Inventory Coverage = Stock Levels / Reorder Point

```python
inventory["inventory_Coverage"] = (
    inventory["Stock_Levels"] /
    inventory["Reorder_Point"]
)
inventory[["Product_ID", "Stock_Levels", "Reorder_Point", "inventory_Coverage"]].head()
```

##### 5. Stockout Risk Analysis :

This KPI identifies products that experience frequent stock shortages.

###### Formula: Stockout Risk = Stockout Frequency


```python
inventory[
    ["Product_ID",
     "Store_ID",
     "Stockout_Frequency"]
].sort_values(
    by="Stockout_Frequency",
    ascending=False
).head()
```

##### 6. Return Rate Analysis :

This KPI measures the percentage of products returned by customers after purchase.

###### Formula: Return Rate = Returned Products / Total Products Sold

```python
pricing[
    ["Product_ID",
     "Sales_Volume",
     "Return_Rate_"]
].sort_values(
    by="Return_Rate_",
    ascending=False
).head()
```

##### 7. Warehouse Capacity Utilization Analysis :
 
This KPI measures how efficiently warehouse space is being used.
 
###### Formula: Capacity Utilization = Stock Levels / Warehouse Capacity

```python
inventory["capacity_utillization"] = (
    inventory["Stock_Levels"] / inventory["Warehouse_Capacity"]

)

inventory[["Product_ID", "Stock_Levels", "Warehouse_Capacity", "capacity_utillization"]].head()
```

##### 8. Procurement Lead Time Analysis :

This KPI measures the time taken by suppliers to deliver products after an order is placed.

###### Formula: Procurement Lead Time = Supplier Delivery Days

```python
inventory[["Product_ID", "Store_ID","Supplier_Lead_Time_"
]].sort_values(
    by="Supplier_Lead_Time_",
    ascending=False
)
```

##### 9. Order Fulfillment Time Analysis :

This KPI measures the average time required to fulfill customer orders.

###### Formula: Fulfillment Time = Order Fulfillment Time (Days)

```python
inventory[["Product_ID", "Store_ID","Order_Fulfillment_Time"]].sort_values(
    by= "Order_Fulfillment_Time",
    ascending=False
).head()
```

##### 10. Customer Review Analysis :
 
This KPI measures customer satisfaction based on product reviews.
 
###### Formula: Average Review Score = Total Customer Reviews / Number of Products

```python
pricing[
    ["Product_ID",
     "Customer_Reviews"]
].sort_values(
    by="Customer_Reviews",
    ascending=False
).head()
```

##### 1. Top 10 Products by Sales

This chart highlights the products with the highest sales quantity.

###### Objective: Identify best-selling products.

```python
import matplotlib.pyplot as plt

top_products = (
    demand.groupby("Product_ID")["Sales_Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(5,3))
top_products.plot(kind="bar", color="#DAA06D")
plt.title("Top 10 Products by Sales")
plt.xlabel("Product ID")
plt.ylabel("Sales Quantity")
plt.show()
```

##### 2. Revenue by Customer Segment

This chart compares revenue generated by each customer segment.

###### Objective: Identify the most valuable customer segment.

```python
segment_revenue = (
    demand.groupby("Customer_Segments")["revenue"]
    .sum()
)

colors = plt.cm.YlOrBr([0.4, 0.6, 0.8])

plt.figure(figsize=(4,4))

plt.pie(
    segment_revenue,
    labels=segment_revenue.index,
    autopct="%1.1f%%",
    colors=colors,
    startangle=90
)

plt.title("Revenue Contribution by Customer Segment")

plt.show()
```

##### 3. Promotion Impact on Sales

This chart analyzes whether promotions influence sales quantity.

###### Objective: Measure promotional effectiveness.

```python
sns.boxplot(
    data=demand,
    x="Promotions_",
    y="Sales_Quantity",
    hue="Promotions_",
    palette=["#EC7F70", "#7AC25E"],
    legend=False
)

plt.title("Promotion Impact on Sales")
plt.xlabel("Promotions")
plt.ylabel("Sales Quantity")
plt.show()
```

##### 4. Seasonal Sales Analysis

This visualization compares average sales across different seasons.

###### Objective: Identify seasons with the highest product demand.

```python
season_sales = (
    demand.groupby("Seasonality_Factors")["Sales_Quantity"]
    .mean()
)

plt.figure(figsize=(6, 4))

plt.pie(
    season_sales,
    labels=season_sales.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=[ "#B87333", "#8B5A2B", "#E6B17E"],
    wedgeprops={"width": 0.4, "edgecolor": "white"}
)

plt.title("Average Sales by Season")

plt.show()
```

##### 5. Stockout Frequency Analysis

This visualization shows the distribution of stockout frequency.

###### Objective: Identify inventory shortage patterns.

```python
plt.figure(figsize=(6,4))

plt.hist(
    inventory["Stockout_Frequency"],
    bins=20,
    color="#CD853F"
)

plt.title("Distribution of Stockout Frequency")
plt.xlabel("Stockout Frequency")
plt.ylabel("Count")

plt.show()
```

##### 6. Monthly Return Rate Analysis

This visualization shows how return rates vary across months.

###### Objective: Identify seasonal patterns in product returns.

```python
merged = pd.merge(
    demand,
    pricing,
    on=["Product_ID", "Product_ID"],
    how="inner"
)
```

```python
merged["Month"] = pd.to_datetime(
    merged["Date_"]
).dt.month_name()

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_returns = (
    merged.groupby("Month")["Return_Rate_"]
    .mean()
    .reindex(month_order)
)

plt.figure(figsize=(9,4))

plt.plot(
    monthly_returns.index,
    monthly_returns.values,
    marker="o",
    linewidth=2,
    color="#926A20"
)

# Data labels
for i, value in enumerate(monthly_returns.values):
    plt.text(
        i,
        value,
        f"{value:.1f}",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.title("Average Return Rate by Month", fontdict={"fontsize": 12, "fontweight": "bold"})
plt.xlabel("Month", fontdict={"fontsize": 10, "fontweight": "bold"})
plt.ylabel("Average Return Rate (%)", fontdict={"fontsize": 10, "fontweight": "bold"})

plt.xticks(fontsize=7)   # Month names
plt.yticks(fontsize=8)   # Y-axis values


plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

##### 10. Lead Time vs Stockout Frequency

This visualization analyzes the relationship between supplier lead time and stockout frequency.

###### Objective: Determine whether longer lead times contribute to inventory shortages.

##### 12. Company Price vs Competitor Price

This visualization compares the price distribution of company products against competitor prices.

###### Objective: Evaluate pricing competitiveness in the market.

```python
avg_company = pricing["Price_"].mean()
avg_competitor = pricing["Competitor_Prices"].mean()

plt.figure(figsize=(5,4))

bars = plt.bar(
    ["Company Price", "Competitor Price"],
    [avg_company, avg_competitor],
    color=["#AAE046", "#E46C57"]
)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():.2f}",
        ha="center"
    )

plt.title("Average Company Price vs Competitor Price")
plt.ylabel("Average Price")

plt.show()
```
